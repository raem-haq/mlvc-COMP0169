# Linear Fitting

A compact supporting experiment in least-squares regression using the Iris dataset. It is kept separate from the image-focused notebooks. The polynomial extension remains unfinished.


```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn import datasets as skdataset
from sklearn.model_selection import train_test_split

RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)

iris = skdataset.load_iris()
X_iris, X_iris_test, Y_iris, Y_iris_test = train_test_split(
    iris.data, iris.target, test_size=0.1, random_state=RANDOM_STATE
)


def plot_correlation(actual, predicted):
    actual = np.asarray(actual).ravel()
    predicted = np.asarray(predicted).ravel()
    plt.figure(figsize=(6, 6))
    plt.scatter(actual, predicted, edgecolor="black")
    lower = min(actual.min(), predicted.min())
    upper = max(actual.max(), predicted.max())
    plt.plot([lower, upper], [lower, upper], color="red")
    plt.xlabel("Actual")
    plt.ylabel("Predicted")
    plt.tight_layout()
    plt.show()
```

## Normal-equation solver


```python
def nsolve(X,y):
    """
    Solve least squares using the Moore-Penrose pseudoinverse.

    Arguments:
    X : Data matrix
    y : Labels

    Returns:
    w : Weights
    """

    w = np.linalg.pinv(X) @ y 

    return w
```

## Synthetic-data check


```python

np.random.seed(RANDOM_STATE)
X = np.random.randn(100, 5)
y = np.random.randint(0,2,size=(100,))

w = nsolve(X, y)

print("Computed weights (w):\n", w)

# Verify by comparing with numpy's least squares solution
w_lstsq, _, _, _ = np.linalg.lstsq(X, y, rcond=None)
print("\nnp.linalg.lstsq weights:\n", w_lstsq)

print("\nAre the two results close?", np.allclose(w, w_lstsq))
```

## Linear regression on Iris


```python
def line_fit(X,y):
    """
    Fit a linear model and report mean squared prediction error.

    Arguments:
    X : Data matrix
    y : Labels

    Returns:
    w : Weights
    l2_error : L2 Prediction error using learned w
    """

    y = y.reshape(-1, 1)
    
    X_aug = np.hstack([X, np.ones((X.shape[0], 1))])
    
    w = nsolve(X_aug, y)
    
    y_pred = X_aug @ w
    
    l2_error = np.mean((y - y_pred) ** 2)

    return w, l2_error
```

```python

#remove petal length (3rd column) to use other features to predict it
X_train = np.delete(X_iris, 2, axis=1)  
y_train = X_iris[:, 2]                  

X_test = np.delete(X_iris_test, 2, axis=1)
y_test = X_iris_test[:, 2]

# Fit model on training set
w, train_error = line_fit(X_train, y_train)
print(f"L2 error on training set: {train_error:.4f}")
```

```python

X_test_aug = np.hstack([X_test, np.ones((X_test.shape[0], 1))])
y_pred_test = X_test_aug @ w
test_error = np.mean((y_test.reshape(-1,1) - y_pred_test) ** 2)
print(f"L2 error on test set: {test_error:.4f}")
```

```python

plot_correlation(y_test, y_pred_test)
```
